```python
import requests as uReq
from bs4 import BeautifulSoup as soup
import numpy as np
import csv

page =1
for page in np.arange(5):
    url = uReq.get("https://www.myanmarbusiness-directory.com/en/categories-index/desktop-publishing-services/page{}.html?city=%E1%80%99%E1%80%82%E1%81%A4%E1%80%9C%E1%80%AC%E1%80%92%E1%80%B6%E1%80%AF%E1%81%BF%E1%80%99%E1%80%AD%E1%80%B3%E1%82%95%E1%80%94%E1%80%9A%E1%80%B9".format(page))
    page = page+1
    page_soup = soup(url.text,"html.parser")
    info = page_soup.findAll("div",{"class: ","row detail_row"})

    
    for data in info :
        title_data = data.findAll("p",{"class: ","title"})
        business = title_data[0].text.strip()
        biz = business.replace("featured"and","," ")
        Business_name = " ".join(biz.split())
        print("Business_name :"+ Business_name)
      
        Add =  data.find_all("p",class_=None)[0]
        Address1 = Add.text
        print("Address1 :"+Address1)
        
        Add2 =  data.find_all("p",class_=None)[1]
        Address2 = Add2.text
        print("Address2 :" + Address2)
   
        Phone =  data.find_all("p",class_=None)[2]
        Phone_no = Phone.text
        print("Phone_no :"+Phone_no)

        filename = "desktop-publishing-services.csv"
        file = open(filename,"a",encoding="utf-8")
 
        field = [(Business_name + ","+ Address1 + ","+ Address2 + ","+ Phone_no)]
        writer = csv.writer(file)
        writer.writerow(field)
        file.close()
    
```
